```python
import pandas as pd
import numpy as np

import seaborn as sns
from matplotlib.cbook import boxplot_stats
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    classification_report, confusion_matrix,
    f1_score, precision_score, recall_score,
    roc_auc_score, ConfusionMatrixDisplay
)
```

### Etapa 1: Carregamento e Filtragem dos Dados

```python
csv_path = "data/US_Accidents_March23.csv"
df = pd.read_csv(csv_path, encoding="latin-1", nrows=1000000)
df.head()
```

```python
df.columns
```

```python
df.shape
```

```python
df.info()
```

```python
threshold = 0.85  # remove colunas com mais de 85% de nulos
df = df.dropna(axis=1, thresh=int(threshold * len(df)))
```

```python
def show_metadata(df):
    rows = []
    for col in df.columns:
        serie = df[col]
        n_total = len(serie)
        n_null = int(serie.isnull().sum())
        pct_null = round(n_null / n_total * 100, 1)
        n_uniq = serie.nunique()
        samples = serie.dropna().head(3).tolist()
        rows.append({
           "Coluna":        col,
            "Tipo":          str(serie.dtype),
            "Nulos (qtd)":   n_null,
            "Nulos (%)":     pct_null,
            "Cardinalidade": n_uniq,
            "Exemplos":      str(samples),
        })
    _metadata = pd.DataFrame(rows).set_index("Coluna")
    return _metadata
```

```python
_metadata = show_metadata(df)
_metadata
```

```python
# Criando variáveis multirotulo a partir da váriavel target (severity)
# Seguindo uma abordagem cumulativa ordinal binária (>=)

df['Severity_lv1'] = (df['Severity'] >= 1).astype(int)
df['Severity_lv2'] = (df['Severity'] >= 2).astype(int)
df['Severity_lv3'] = (df['Severity'] >= 3).astype(int)
df['Severity_lv4'] = (df['Severity'] >= 4).astype(int)

multilabel_targets = ['Severity_lv2', 'Severity_lv3', 'Severity_lv4']  # lv1 é trivial (todos = 1), pode omitir
df.head()
```

```python
# Removendo colunas com somente 1 cardinalidade
cols_to_drop = [col for col in df.columns if df[col].nunique() == 1]
print("Removendo:", cols_to_drop)
df = df.drop(columns=cols_to_drop)
```

```python
df.describe()
```

### Etapa 2: Preprocessamento dos Dados

```python
# Transformando os dados temporais para tipos de data
df["Start_Time"] = pd.to_datetime(df["Start_Time"])
df["Month"] = df["Start_Time"].dt.month
df["Day"] = df["Start_Time"].dt.day
df["Hour"] = df["Start_Time"].dt.hour
df["Weekday"] = df["Start_Time"].dt.weekday
```

```python
df[["Start_Time", "Month", "Day", "Hour", "Weekday"]].head()
```

```python
features = [
    "Start_Lat", "Start_Lng",            # Traffic (localização)
    "Distance(mi)",                      # Traffic (extensão)
    "Month", "Day", "Hour", "Weekday",   # Temporal
    "Pressure(in)", "Temperature(F)",    # Weather
    "Humidity(%)", "Visibility(mi)",     # Weather
    "Traffic_Signal", "Junction",        # POI
    "Crossing", "Stop",                  # POI
]
target = multilabel_targets
df = df[features + target].copy()
df.head()
```

```python
_metadata = show_metadata(df)
_metadata
```

```python
col_numerico_ausentes = ["Pressure(in)", "Temperature(F)", "Humidity(%)", "Visibility(mi)"]
df[col_numerico_ausentes].describe()
```

```python
ax = sns.boxplot(data=df[col_numerico_ausentes])
```

```python
# Imputando a mediana nos valores ausentes 
# nas colunas numericas com dados ausentes
for col in col_numerico_ausentes:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())
```

```python
boxplot_stats(df[col_numerico_ausentes])
```

```python
for col in col_numerico_ausentes:
    outliers = boxplot_stats(df[col]).pop(0)['fliers']
    mask = df[col].isin(outliers)
    n = mask.sum()
    df.drop(df.index[mask], inplace=True)
    print(f"{n} Outliers removidos na coluna {col} com os seguintes valores: {outliers}")
```

```python
ax = sns.boxplot(data=df[col_numerico_ausentes])
```

```python
_metadata = show_metadata(df)
_metadata
```

```python
# distribução da variavel target (desbalanceada)
print("=" * 55)
print("Dataset US-Accidents - Estrutura")
print("=" * 55)
print(f"Instâncias: {df[features].shape[0]}")
print(f"Features: {df[features].shape[1]}")
print(f"Rótulos: {df[target].shape[1]}")
print("Distribuição dos rótulos:")
for i, label_name in enumerate(target):
    freq = df[label_name].mean() * 100
    print(f"  {label_name:<22}: {freq:.1f}% positivos")

# dist = df[target].value_counts().sort_index()
# print(f"\n  Distribuição das classes (Severity):")
# for cls, cnt in dist.items():
#     pct = cnt / len(df) * 100
#     print(f"    Nível {cls}: {cnt:,} ({pct:.2f}%)")
```

### Divisão do frame em Treino e Teste

```python
X = df[features]
y = df[target]
X.shape, y.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
X_train.shape, X_test.shape, y_train.shape, y_test.shape 
```

```python
# Normalização
scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)
X_test_sc = scaler.transform(X_test)
```

```python
X_train_sc
```

### Treinamento do Modelo

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)
```

### Validação do modelo

```python
from sklearn.metrics import multilabel_confusion_matrix, classification_report

target_columns = y_test.columns
y_pred = knn_model.predict(X_test)

# Matriz de confusão para cada label
mcm = multilabel_confusion_matrix(y_test, y_pred)

for i, label in enumerate(target_columns):
    tn, fp, fn, tp = mcm[i].ravel()
    print(f"\nLabel {label}:")
    print(f"    TP: {tp}, TN: {tn}, FP: {fp}, FN: {fn}")
    
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    
    print(f"    Precision: {precision:.4f}")
    print(f"    Recall:    {recall:.4f}")
    print(f"    F1-Score:  {f1:.4f}")
```

### Testando diferentes valores para k

```python
from sklearn.multioutput import MultiOutputClassifier
from sklearn.metrics import f1_score, accuracy_score, hamming_loss, jaccard_score

k_values = [1, 3, 5, 7, 9]
f1_scores_macro = []
f1_scores_micro = []
f1_scores_samples = []
hamming_losses = []
exact_match_acc = []

for k in k_values:
    # Para multilabel, use MultiOutputClassifier
    base_knn = KNeighborsClassifier(n_neighbors=k)
    knn = MultiOutputClassifier(base_knn)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    
    # predict_proba retorna uma lista de arrays (um por target)
    # y_prob = knn.predict_proba(X_test)  # Comentado pois pode ser pesado
    
    # Métricas multilabel
    f1_macro = f1_score(y_test, y_pred, average="macro", zero_division=0)
    f1_micro = f1_score(y_test, y_pred, average="micro", zero_division=0)
    f1_samples = f1_score(y_test, y_pred, average="samples", zero_division=0)
    hamming = hamming_loss(y_test, y_pred)
    exact_match = accuracy_score(y_test, y_pred)  # Todas as labels corretas
    
    f1_scores_macro.append(f1_macro)
    f1_scores_micro.append(f1_micro)
    f1_scores_samples.append(f1_samples)
    hamming_losses.append(hamming)
    exact_match_acc.append(exact_match)
    
    print(f"K={k:2d}: F1(macro)={f1_macro:.4f} | F1(micro)={f1_micro:.4f} | F1(samples)={f1_samples:.4f} | Hamming={hamming:.4f}")

# Resultados
print("\n" + "=" * 70)
print("RESUMO DOS RESULTADOS")
print("=" * 70)
print(f"{'K':<4} {'F1-Macro':<10} {'F1-Micro':<10} {'F1-Samples':<12} {'Hamming':<10} {'ExactMatch':<12}")
print("-" * 70)
for i, k in enumerate(k_values):
    print(f"{k:<4} {f1_scores_macro[i]:<10.4f} {f1_scores_micro[i]:<10.4f} {f1_scores_samples[i]:<12.4f} {hamming_losses[i]:<10.4f} {exact_match_acc[i]:<12.4f}")

# Melhor K baseado em diferentes métricas
best_k_macro = k_values[np.argmax(f1_scores_macro)]
best_k_micro = k_values[np.argmax(f1_scores_micro)]
best_k_hamming = k_values[np.argmin(hamming_losses)]

print("\n" + "=" * 70)
print("MELHORES PARÂMETROS")
print("=" * 70)
print(f"Melhor K para F1-Macro: {best_k_macro} (F1={max(f1_scores_macro):.4f})")
print(f"Melhor K para F1-Micro: {best_k_micro} (F1={max(f1_scores_micro):.4f})")
print(f"Melhor K para Hamming Loss: {best_k_hamming} (Loss={min(hamming_losses):.4f})")
```
