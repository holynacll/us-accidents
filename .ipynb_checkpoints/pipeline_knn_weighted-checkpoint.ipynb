```python
import time

import pandas as pd
import numpy as np

import seaborn as sns
from matplotlib.cbook import boxplot_stats
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    classification_report, confusion_matrix,
    f1_score, precision_score, recall_score,
    roc_auc_score, ConfusionMatrixDisplay
)
```

### Etapa 1: Carregamento e Filtragem dos Dados

```python
csv_path = "data/US_Accidents_March23.csv"
df = pd.read_csv(csv_path, encoding="latin-1", nrows=300000)
df.head()
```

```python
df.columns
```

```python
df.shape
```

```python
df.info()
```

```python
threshold = 0.85  # remove colunas com mais de 85% de nulos
df = df.dropna(axis=1, thresh=int(threshold * len(df)))
```

```python
def show_metadata(df):
    rows = []
    for col in df.columns:
        serie = df[col]
        n_total = len(serie)
        n_null = int(serie.isnull().sum())
        pct_null = round(n_null / n_total * 100, 1)
        n_uniq = serie.nunique()
        samples = serie.dropna().head(3).tolist()
        rows.append({
           "Coluna":        col,
            "Tipo":          str(serie.dtype),
            "Nulos (qtd)":   n_null,
            "Nulos (%)":     pct_null,
            "Cardinalidade": n_uniq,
            "Exemplos":      str(samples),
        })
    _metadata = pd.DataFrame(rows).set_index("Coluna")
    return _metadata
```

```python
_metadata = show_metadata(df)
_metadata
```

```python
# Removendo colunas com somente 1 cardinalidade
cols_to_drop = [col for col in df.columns if df[col].nunique() == 1]
print("Removendo:", cols_to_drop)
df = df.drop(columns=cols_to_drop)
```

```python
df.describe()
```

```python
# Filtrar severidade por 1, 2 e 3 (o artigo usa esses 3 niveis)
print(f"  Antes do filtro Severity = {{1,2,3}}: {df.shape[0]:,} registros")
df = df[df["Severity"].isin([1,2,3])]
print(f"  Após filtro Severity = {{1,2,3}}: {df.shape[0]:,} registros")
```

### Etapa 2: Preprocessamento dos Dados

```python
# Transformando os dados temporais para tipos de data
df["Start_Time"] = pd.to_datetime(df["Start_Time"])
df["Month"] = df["Start_Time"].dt.month
df["Day"] = df["Start_Time"].dt.day
df["Hour"] = df["Start_Time"].dt.hour
df["Weekday"] = df["Start_Time"].dt.weekday
```

```python
df[["Start_Time", "Month", "Day", "Hour", "Weekday"]].head()
```

```python
# Convertendo Booleans em Int
df['Traffic_Signal'] = df['Traffic_Signal'].astype(int)
df['Junction'] = df['Junction'].astype(int)
df['Crossing'] = df['Crossing'].astype(int)
df['Stop'] = df['Stop'].astype(int)
```

```python
features = [
    "Start_Lat", "Start_Lng",            # Traffic (localização)
    "Distance(mi)",                      # Traffic (extensão)
    "Month", "Day", "Hour", "Weekday",   # Temporal
    "Pressure(in)", "Temperature(F)",    # Weather
    "Humidity(%)", "Visibility(mi)",     # Weather
    "Traffic_Signal", "Junction",        # POI
    "Crossing", "Stop",                  # POI
]
target = "Severity"
df = df[features + [target]].copy()
df.head()
```

```python
_metadata = show_metadata(df)
_metadata
```

```python
col_numerico_ausentes = ["Pressure(in)", "Temperature(F)", "Humidity(%)", "Visibility(mi)"]
df[col_numerico_ausentes].describe()
```

```python
ax = sns.boxplot(data=df[col_numerico_ausentes])
```

```python
# Imputando a mediana nos valores ausentes 
# nas colunas numericas com dados ausentes
for col in col_numerico_ausentes:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())
```

```python
boxplot_stats(df[col_numerico_ausentes])
```

```python
for col in col_numerico_ausentes:
    outliers = boxplot_stats(df[col]).pop(0)['fliers']
    mask = df[col].isin(outliers)
    n = mask.sum()
    df.drop(df.index[mask], inplace=True)
    print(f"{n} Outliers removidos na coluna {col} com os seguintes valores: {outliers}")
```

```python
ax = sns.boxplot(data=df[col_numerico_ausentes])
```

```python
_metadata = show_metadata(df)
_metadata
```

```python
# distribução da variavel target (desbalanceada)
dist = df[target].value_counts().sort_index()
print(f"\n  Distribuição das classes (Severity):")
for cls, cnt in dist.items():
    pct = cnt / len(df) * 100
    print(f"    Nível {cls}: {cnt:,} ({pct:.2f}%)")
```

### Divisão do frame em Treino e Teste

```python
X = df[features]
y = df[target]
X.shape, y.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
X_train.shape, X_test.shape, y_train.shape, y_test.shape 
```

```python
# Normalização
scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)
X_test_sc = scaler.transform(X_test)
```

### Treinamento do Modelo

```python
knn_model = KNeighborsClassifier(n_neighbors=5, weights="uniform", n_jobs=-1)
knn_model.fit(X_train_sc, y_train)
```

### Validação do modelo

#### Função auxiliar: avaliação de modelo

```python
def avaliar_modelo(nome, y_true, y_pred, y_prob, classes):
    """Calcula e exibe métricas para modelos multiclass desbalanceados."""
    precision = precision_score(y_true, y_pred, average="macro", zero_division=0)
    recall    = recall_score(y_true, y_pred, average="macro", zero_division=0)
    f1_mac    = f1_score(y_true, y_pred, average="macro", zero_division=0)
    f1_wt     = f1_score(y_true, y_pred, average="weighted", zero_division=0)

    print(f"\n{'='*55}")
    print(f"  Modelo: {nome}")
    print(f"{'='*55}")
    print(f"  Precision (macro):   {precision:.4f}")
    print(f"  Recall    (macro):   {recall:.4f}")
    print(f"  F1        (macro):   {f1_mac:.4f}")
    print(f"  F1        (weighted):{f1_wt:.4f}")
    print()
    print(classification_report(y_true, y_pred,
                                target_names=[f"Nível {c}" for c in classes],
                                zero_division=0))
    return {"modelo": nome, "precision_macro": precision, "recall_macro": recall,
            "f1_macro": f1_mac, "f1_weighted": f1_wt}
```

### 1. KNN Clássico — k=5, weights='uniform' (dados normalizados)

```python
# KNN Clássico com dados normalizados (X_train_sc / X_test_sc)
classes = sorted(y_test.unique())

knn_uniform = KNeighborsClassifier(n_neighbors=5, weights="uniform", n_jobs=-1)

inicio = time.time() 

knn_uniform.fit(X_train_sc, y_train)
y_pred_uni  = knn_uniform.predict(X_test_sc)
y_prob_uni  = knn_uniform.predict_proba(X_test_sc)

duracao = time.time() - inicio
print(f"Treinamento concluido - Tempo: {duracao:.2f}s")

res_uniform = avaliar_modelo("KNN Clássico (uniform, k=5)", y_test, y_pred_uni, y_prob_uni, classes)
```

---
## KNN Ponderado (Weighted KNN) vs KNN Clássico

O KNN ponderado usa `weights='distance'`, atribuindo maior peso aos vizinhos mais próximos. Isso é especialmente útil em datasets desbalanceados, onde a classe majoritária pode dominar a votação simples.

### 2. KNN Ponderado — k=5, weights='distance' (dados normalizados)

```python
# KNN Ponderado: vizinhos mais próximos recebem mais peso na votação
knn_weighted = KNeighborsClassifier(n_neighbors=5, weights="distance", n_jobs=-1)

inicio = time.time() 

knn_weighted.fit(X_train_sc, y_train)
y_pred_wt = knn_weighted.predict(X_test_sc)
y_prob_wt = knn_weighted.predict_proba(X_test_sc)

duracao = time.time() - inicio
print(f"Treinamento concluido - Tempo: {duracao:.2f}s")
    
res_weighted = avaliar_modelo("KNN Ponderado (distance, k=5)", y_test, y_pred_wt, y_prob_wt, classes)
```

### 3. Varredura de K — Clássico vs Ponderado

```python
k_values = [1, 3, 5, 7]
resultados_k = []

for k in k_values:
    for w, label in [("uniform", "Clássico"), ("distance", "Ponderado")]:
        inicio = time.time()
        
        knn = KNeighborsClassifier(n_neighbors=k, weights=w, n_jobs=-1)
        knn.fit(X_train_sc, y_train)
        yp  = knn.predict(X_test_sc)
        ypr = knn.predict_proba(X_test_sc)
        f1  = f1_score(y_test, yp, average="macro", zero_division=0)

        duracao = time.time() - inicio
        
        resultados_k.append({"k": k, "tipo": label, "f1_macro": f1})
        print(f"  k={k:2d} | {label:10s}: F1={f1:.4f} | Tempo={duracao:.2f}s")

df_k = pd.DataFrame(resultados_k)
print("\nConcluído!")
```

### 4. Visualização: F1-macro

```python
fig, ax = plt.subplots(1, 1, figsize=(14, 5))

metric = "f1_macro"
title = "F1-Score (macro)"

for tipo, grp in df_k.groupby("tipo"):
    ax.plot(grp["k"], grp[metric], marker="o", label=tipo)

ax.set_title(title)
ax.set_xlabel("k (vizinhos)")
ax.set_ylabel(metric)
ax.legend()
ax.grid(True, linestyle="--", alpha=0.5)

plt.suptitle("KNN Clássico vs KNN Ponderado — Dataset Desbalanceado", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()
```

### 5. Matrizes de Confusão — k=5

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
labels = [f"Nível {c}" for c in classes]

for ax, y_pred, titulo in zip(
    axes,
    [y_pred_uni, y_pred_wt],
    ["KNN Clássico (uniform, k=5)", "KNN Ponderado (distance, k=5)"]
):
    cm = confusion_matrix(y_test, y_pred, labels=classes)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=labels)
    disp.plot(ax=ax, colorbar=False, cmap="Blues")
    ax.set_title(titulo)

plt.suptitle("Matrizes de Confusão — Comparação KNN", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()
```

### 6. Tabela Comparativa Final

```python
df_comp = pd.DataFrame([res_uniform, res_weighted])
df_comp = df_comp.set_index("modelo")
df_comp.columns = ["Precision (macro)", "Recall (macro)", "F1 (macro)", "F1 (weighted)"]
df_comp.style.highlight_max(axis=0, color="lightgreen").format("{:.4f}")
```
